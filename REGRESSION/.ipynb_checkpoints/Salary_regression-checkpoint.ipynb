```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('data/Salary Data.csv')
df.head()
```

```python
plt.scatter(df['Years of Experience'],df.Salary,color='red')
plt.show()
```

```python
df.hist()
```

```python
cdf = df[['Years of Experience','Salary']]
cdf["Salary"] = cdf["Salary"].fillna(cdf["Salary"].median())
cdf["Years of Experience"] = cdf["Years of Experience"].fillna(cdf["Years of Experience"].median())
```

```python
df.dropna()
msk = np.random.rand(len(cdf['Salary'])) < 0.8

train = cdf[msk]
test = cdf[~msk]
```

```python
from sklearn import linear_model
reg = linear_model.LinearRegression()
trian_x = np.asanyarray(train[['Years of Experience']])
train_y = np.asanyarray(train[['Salary']])
reg.fit(trian_x,train_y)
print(reg.coef_)
print(reg.intercept_)
```
